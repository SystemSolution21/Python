```python
# Fibonacci Implementations

## 1. **Naive Recursive**: O(2ⁿ) time, O(n) space
def fib_naive(n):
    if n <= 1:
        return n
    return fib_naive(n - 1) + fib_naive(n - 2)


print(fib_naive(10))
```

```python
##2. **Memoized (Top-Down)**: O(n) time, O(n) space

from functools import cache


@cache
def fib_memo(n: int) -> int:
    if n <= 1:
        return n
    return fib_memo(n - 1) + fib_memo(n - 2)


print(fib_memo(10))
```

```python
## 3. **Tabulation (Bottom-Up)**: O(n) time, O(n) space
def fibonacci_tab(n: int) -> int:
    """
    Calculate the nth Fibonacci number using tabulation.

    Args:
        n: The position in the Fibonacci sequence

    Returns:
        The nth Fibonacci number
    """
    if n <= 1:
        return n

    dp: list[int] = [0] * (n + 1)
    dp[1] = 1

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]


n: int = 100
print(f"{n}th Fibonacci number: {fibonacci_tab(n=n)}")
```

```python
## 4. **Optimized Iterative**: O(n) time, O(1) space


def fib_optimized(n: int) -> int:
    previous, current = 0, 1
    for _ in range(n):
        previous, current = current, previous + current
    return previous


print(fib_optimized(10))
```

```python
## 5. **Hash-map solution**: O(n) time, O(n) space


def find_two_numbers_with_sum(
    numbers: list[int], target_sum: int
) -> tuple[int, int] | None:
    seen_indices: dict[int, int] = {}

    for index, current_number in enumerate(iterable=numbers):
        complement: int = target_sum - current_number
        if complement in seen_indices:
            return seen_indices[complement], index

        seen_indices[current_number] = index

    return None


numbers: list[int] = [2, 7, 11, 15, 1]
target_sum: int = 22

pair_indices: tuple[int, int] | None = find_two_numbers_with_sum(
    numbers=numbers, target_sum=target_sum
)
if pair_indices is None:
    print("No two numbers add up to the target sum.")
else:
    first_index, second_index = pair_indices
    print(f"{target_sum} = {numbers[first_index]} + {numbers[second_index]}")
```

```python
## 6. **Fibonacci Sequence Generator**:
## Time: O(k), where k is the number of Fibonacci numbers generated until the limit is reached.
## Space: O(k) for the list that stores the sequence


def generate_fibonacci_sequence(limit: int) -> list[int]:
    """Return all Fibonacci numbers smaller than the given limit."""
    current, next_value = 0, 1
    sequence: list[int] = []

    while current < limit:
        sequence.append(current)
        current, next_value = next_value, current + next_value

    return sequence


limit: int = 1000
result: list[int] = generate_fibonacci_sequence(limit=limit)
print(f"{len(result)} Fibonacci Sequence: {result} generated from given limit {limit}.")
```
