```python
def check_input(text: str) -> tuple[str, str]:
    get_len: int = len(text)
    get_char: str = ""
    get_num: str = ""
    for i in range(get_len):
        if text[i] in "0123456789":
            get_num += text[i]
        else:
            get_char += text[i]
    return get_char, get_num


def main() -> None:
    while True:
        text = input("Please enter any, or type `exit` to quit. ").strip()
        if not text:
            continue

        if text.lower() == "exit":
            break

        result: tuple[str, str] = check_input(text=text)
        print(result)


if __name__ == "__main__":
    main()
```
